```python
import numpy as np
from reion.models import GenLogistic

# Choose thesis parameters
z_mid = 7.5
Delta = 0.5

# Convert to the new parameterisation
dz2575 = 2 * np.log(3) * Delta

# Create Vikram's model with nu = 1
model = GenLogistic(
    z50=z_mid,
    dz2575=dz2575,
    nu=1.0
)

# Check parameter recovery
print("z0 =", model.z0)
print("D =", model.D)

# Compare Q_HII at several redshifts
z = np.array([6.0, 7.0, 7.5, 8.0, 9.0])

Q_thesis = 1 / (1 + np.exp((z - z_mid) / Delta))
Q_general = model.Q(z)

print("\nThesis logistic:", Q_thesis)
print("GenLogistic:    ", Q_general)
print("Maximum difference:", np.max(np.abs(Q_thesis - Q_general)))
```

```python
import numpy as np
from reion.likelihood import XHIData
from reion.models import GenLogistic

# Load the Gold logistic posterior
post = np.load("results/timing_logistic_gold.npz")["post"]

# Calculate posterior median parameters
med = np.median(post, axis=0)

# Construct the logistic model
model = GenLogistic(
    med[0],
    np.exp(med[1]),
    1.0
)

# Load the extended dataset
data = XHIData("data/xhi_constraints.csv", subset="extended")

# Calculate model predictions and individual likelihoods
data.df["x_model"] = model.xHI(data.z)
data.df["lnL"] = data.lnlike_terms(model.xHI(data.z))

# Display the six problematic rows at full precision
ids = [
    "Morales21_z73",
    "Umeda25_z57_LF",
    "Hoag19_LBG",
    "Morales21_z70",
    "burton23_gnz11",
    "Morales21_z66"
]

print(
    data.df[data.df["id"].isin(ids)]
    [["id", "z", "x", "ep", "em", "kind",
      "probe", "x_model", "lnL"]]
    .to_string(index=False)
)
```
